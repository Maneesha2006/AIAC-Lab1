task1:With AI assistance, create a Vehicle Service Management database.
•	Include tables: Customers, Vehicles, Service_Records.
•	Insert Data in to these three tables
•	Write a query to find all vehicles serviced in the last 30 days


```python
import sqlite3
import datetime
from tabulate import tabulate  # for pretty table output

# --- 1. Establish connection to an in-memory SQLite database ---
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Enable foreign key constraints
cursor.execute('PRAGMA foreign_keys = ON;')

print("Connected to in-memory SQLite database and foreign keys enabled.\n")

# --- 2. Define SQL CREATE TABLE statements (SQLite compatible) ---

create_customers_table_sql = """
CREATE TABLE IF NOT EXISTS Customers (
    CustomerID INTEGER PRIMARY KEY AUTOINCREMENT,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    Email TEXT,
    Phone TEXT,
    CreatedAt TEXT NOT NULL DEFAULT (CURRENT_TIMESTAMP)
);
"""

create_vehicles_table_sql = """
CREATE TABLE IF NOT EXISTS Vehicles (
    VehicleID INTEGER PRIMARY KEY AUTOINCREMENT,
    CustomerID INTEGER NOT NULL,
    Make TEXT,
    Model TEXT,
    Year INTEGER,
    VIN TEXT UNIQUE,
    CreatedAt TEXT NOT NULL DEFAULT (CURRENT_TIMESTAMP),
    FOREIGN KEY (CustomerID) REFERENCES Customers(CustomerID)
);
"""

create_service_records_table_sql = """
CREATE TABLE IF NOT EXISTS Service_Records (
    ServiceID INTEGER PRIMARY KEY AUTOINCREMENT,
    VehicleID INTEGER NOT NULL,
    ServiceDate TEXT NOT NULL,
    Description TEXT,
    ServiceCost REAL,
    Technician TEXT,
    CreatedAt TEXT NOT NULL DEFAULT (CURRENT_TIMESTAMP),
    FOREIGN KEY (VehicleID) REFERENCES Vehicles(VehicleID)
);
"""

# --- 3. Execute CREATE TABLE statements ---
cursor.execute(create_customers_table_sql)
cursor.execute(create_vehicles_table_sql)
cursor.execute(create_service_records_table_sql)
conn.commit()
print("Tables 'Customers', 'Vehicles', and 'Service_Records' created successfully.\n")

# --- 4. Define and Insert Sample Data ---

# Sample data for Customers
customers_data = [
    ('John', 'Doe', 'john.doe@example.com', '555-0100'),
    ('Jane', 'Smith', 'jane.smith@example.com', '555-0200'),
    ('Robert', 'Lee', 'robert.lee@example.com', '555-0300')
]
cursor.executemany("INSERT INTO Customers (FirstName, LastName, Email, Phone) VALUES (?, ?, ?, ?)", customers_data)

# Retrieve CustomerIDs for Vehicle data
cursor.execute("SELECT CustomerID FROM Customers ORDER BY CustomerID ASC;")
customer_ids = [row[0] for row in cursor.fetchall()]

# Sample data for Vehicles
vehicles_data = [
    (customer_ids[0], 'Toyota', 'Camry', 2015, 'VIN001'),
    (customer_ids[1], 'Honda', 'Civic', 2018, 'VIN002'),
    (customer_ids[0], 'Ford', 'F-150', 2020, 'VIN003'),
    (customer_ids[2], 'Tesla', 'Model 3', 2022, 'VIN004')
]
cursor.executemany("INSERT INTO Vehicles (CustomerID, Make, Model, Year, VIN) VALUES (?, ?, ?, ?, ?)", vehicles_data)

# Retrieve VehicleIDs for Service_Records data
cursor.execute("SELECT VehicleID FROM Vehicles ORDER BY VehicleID ASC;")
vehicle_ids = [row[0] for row in cursor.fetchall()]

# Sample data for Service_Records (dates calculated relative to 'today')
today = datetime.date.today()
last_30_days = today - datetime.timedelta(days=30)
last_15_days = today - datetime.timedelta(days=15)
older_date = today - datetime.timedelta(days=60)

service_records_data = [
    (vehicle_ids[0], last_15_days.strftime('%Y-%m-%d'), 'Oil change', 75.00, 'Alice'),
    (vehicle_ids[1], last_30_days.strftime('%Y-%m-%d'), 'Tire rotation', 50.00, 'Bob'),
    (vehicle_ids[2], older_date.strftime('%Y-%m-%d'), 'Brake replacement', 450.00, 'Charlie'),
    (vehicle_ids[0], today.strftime('%Y-%m-%d'), 'Engine check', 120.00, 'Alice'),
    (vehicle_ids[3], last_15_days.strftime('%Y-%m-%d'), 'Software update', 0.00, 'David')
]
cursor.executemany("INSERT INTO Service_Records (VehicleID, ServiceDate, Description, ServiceCost, Technician) VALUES (?, ?, ?, ?, ?)", service_records_data)
conn.commit()
print("Sample data inserted successfully.\n")

# --- 5. Query for vehicles serviced in the last 30 days ---
date_30_days_ago_str = last_30_days.strftime('%Y-%m-%d')

query_recent_services = f"""
SELECT DISTINCT
    V.VehicleID,
    V.Make,
    V.Model,
    V.Year,
    SR.ServiceDate,
    SR.Description,
    SR.ServiceCost,
    SR.Technician
FROM Vehicles V
JOIN Service_Records SR ON V.VehicleID = SR.VehicleID
WHERE SR.ServiceDate >= '{date_30_days_ago_str}'
ORDER BY SR.ServiceDate DESC;
"""

print(f"Vehicles serviced in the last 30 days (since {date_30_days_ago_str}):\n")

cursor.execute(query_recent_services)
recent_services = cursor.fetchall()

# --- 6. Display results in table format ---
if recent_services:
    headers = ["VehicleID", "Make", "Model", "Year", "ServiceDate", "Description", "ServiceCost", "Technician"]
    print(tabulate(recent_services, headers=headers, tablefmt="grid"))
else:
    print("No vehicles serviced in the last 30 days.")

# Optional: Close connection
conn.close()
```

Use AI to translate a JAVA array sorting program into Python, and verify that both outputs match for a given array.

```python
public class task2 {
    public static void main(String[] args) {
        int[] original = {5, 3, 8, 1, 2};
        int[] sorted = original.clone();
        java.util.Arrays.sort(sorted);
        System.out.println("Original: " + toPythonListString(original));
        System.out.println("Sorted:   " + toPythonListString(sorted));
    }

    private static String toPythonListString(int[] arr) {
        StringBuilder sb = new StringBuilder();
        sb.append("[");
        for (int i = 0; i < arr.length; i++) {
            sb.append(arr[i]);
            if (i < arr.length - 1) sb.append(", ");
        }
        sb.append("]");
        return sb.toString();
    }
}
```
